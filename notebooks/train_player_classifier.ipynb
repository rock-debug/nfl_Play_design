```python
from engine.player_classifier import PlayerPositionClassifier
from engine.features_player import extract_player_features
import numpy as np

clf = PlayerPositionClassifier()

feature_list = []
label_list = []

# Example: build synthetic training from your engine formations
from engine.formations import singleback_11
from engine.defensive_shells import cover2

off = singleback_11()
defn = cover2()

# use only pre-snap positions for now
for p in off + defn:
    route = np.tile(p.start_pos, (5,2))  # synthetic tiny route
    features = extract_player_features(route)
    feature_list.append(features)
    label_list.append(p.role)

clf.train(feature_list, label_list)
clf.save()
```
